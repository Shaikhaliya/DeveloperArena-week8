### Import Libraries

```python
import pandas as pd
import numpy as np
```

### Load Dataset

```python
df = pd.read_csv("../data/customer_churn.csv")

print(df.head())
```

### Data Exploration

```python
print("========== DATASET INFORMATION ==========")

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

### Data Validation

```python


print("========== DATA VALIDATION ==========")

print("\nTenure Range:")
print("Minimum:", df["Tenure"].min())
print("Maximum:", df["Tenure"].max())

print("\nMonthly Charges Range:")
print("Minimum:", df["MonthlyCharges"].min())
print("Maximum:", df["MonthlyCharges"].max())

print("\nTotal Charges Range:")
print("Minimum:", df["TotalCharges"].min())
print("Maximum:", df["TotalCharges"].max())

print("\nChurn Values:")
print(df["Churn"].value_counts())

print("\nSenior Citizen Values:")
print(df["SeniorCitizen"].value_counts())
```

```python
# Categorical Data Validation

categorical_columns = [
    "Contract",
    "PaymentMethod",
    "PaperlessBilling"
]

for column in categorical_columns:
    print(f"\n========== {column} ==========")
    print(df[column].value_counts())
```

```python

# Create Clean Dataset


cleaned_df = df.copy()

print("Cleaned Dataset Shape:")
print(cleaned_df.shape)
```

```python


print("========== FINAL DATA QUALITY CHECK ==========")

print("\nMissing Values:")
print(cleaned_df.isnull().sum())

print("\nDuplicate Rows:")
print(cleaned_df.duplicated().sum())

print("\nDataset Shape:")
print(cleaned_df.shape)
```

```python


cleaned_df.to_csv(
    "../data/cleaned_data.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

### Conclusion

The dataset was successfully validated and prepared for analysis.

- No missing values were found.
- No duplicate rows were found.
- Numerical data ranges were validated.
- Categorical values were checked.
- The dataset contains 500 customer records.
- The cleaned dataset was saved as `cleaned_data.csv`.

The dataset is ready for exploratory data analysis.
