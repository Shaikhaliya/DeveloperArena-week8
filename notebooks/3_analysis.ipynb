### Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
```

### Load Cleaned Dataset

```python

df = pd.read_csv("../data/cleaned_data.csv")

print(df.head())
print("\nDataset Shape:", df.shape)
```

### Descriptive Statistics by Churn Status

```python


churn_statistics = (
    df.groupby("Churn")[
        ["Tenure", "MonthlyCharges", "TotalCharges"]
    ]
    .agg(["mean", "median", "std", "min", "max"])
)

print("========== DESCRIPTIVE STATISTICS BY CHURN ==========")
print(churn_statistics)
```

### Business Insight: Customer Characteristics by Churn

The descriptive analysis shows a major difference in customer tenure between churned and non-churned customers.

Non-churned customers have an average tenure of approximately 40.15, while churned customers have an average tenure of only 6.00. This suggests that customers are more likely to churn during the early stages of their relationship with the company.

Churned customers have higher average monthly charges (129.77) compared to non-churned customers (111.72). This suggests that monthly charges may be associated with customer churn and should be investigated further using hypothesis testing.

The average total charges are relatively similar between churned and non-churned customers. Further statistical testing will determine whether the observed differences are statistically significant.


### Hypothesis Test 1: Tenure vs Churn


```python

churned_tenure = df[df["Churn"] == 1]["Tenure"]
non_churned_tenure = df[df["Churn"] == 0]["Tenure"]

t_statistic, p_value = stats.ttest_ind(
    churned_tenure,
    non_churned_tenure,
    equal_var=False
)

print("========== HYPOTHESIS TEST 1 ==========")
print("Test: Tenure Difference by Churn Status")
print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Statistically Significant")
    print("Reject the Null Hypothesis.")
else:
    print("Result: Not Statistically Significant")
    print("Fail to Reject the Null Hypothesis.")
```

## Hypothesis Test 1: Tenure and Customer Churn

### Hypotheses

- Null Hypothesis (H₀): The average tenure is the same for churned and non-churned customers.
- Alternative Hypothesis (H₁): The average tenure is different for churned and non-churned customers.

### Result

The p-value is significantly smaller than the significance level of 0.05.

Therefore, the null hypothesis is rejected.

There is a statistically significant difference in customer tenure between churned and non-churned customers.

This finding supports the earlier exploratory analysis, which showed that churned customers have substantially lower average tenure. Customer tenure is therefore an important factor associated with churn in this dataset.

### Hypothesis Test 2: Monthly Charges vs Churn

```python


churned_charges = df[df["Churn"] == 1]["MonthlyCharges"]
non_churned_charges = df[df["Churn"] == 0]["MonthlyCharges"]

t_statistic, p_value = stats.ttest_ind(
    churned_charges,
    non_churned_charges,
    equal_var=False
)

print("========== HYPOTHESIS TEST 2 ==========")
print("Test: Monthly Charges Difference by Churn Status")
print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Statistically Significant")
    print("Reject the Null Hypothesis.")
else:
    print("Result: Not Statistically Significant")
    print("Fail to Reject the Null Hypothesis.")
```

## Hypothesis Test 2: Monthly Charges and Customer Churn

### Hypotheses

- Null Hypothesis (H₀): The average Monthly Charges are the same for churned and non-churned customers.
- Alternative Hypothesis (H₁): The average Monthly Charges are different for churned and non-churned customers.

### Result

The p-value of approximately 0.0101 is smaller than the significance level of 0.05.

Therefore, the null hypothesis is rejected.

There is a statistically significant difference in Monthly Charges between churned and non-churned customers.

The descriptive analysis showed that churned customers have higher average Monthly Charges (129.77) compared to non-churned customers (111.72).

This suggests that Monthly Charges are associated with customer churn in this dataset. However, this statistical relationship does not prove that higher Monthly Charges directly cause customer churn.

### Hypothesis Test 3: Total Charges vs Churn

```python


churned_total_charges = df[df["Churn"] == 1]["TotalCharges"]
non_churned_total_charges = df[df["Churn"] == 0]["TotalCharges"]

t_statistic, p_value = stats.ttest_ind(
    churned_total_charges,
    non_churned_total_charges,
    equal_var=False
)

print("========== HYPOTHESIS TEST 3 ==========")
print("Test: Total Charges Difference by Churn Status")
print("T-statistic:", t_statistic)
print("P-value:", p_value)

if p_value < 0.05:
    print("Result: Statistically Significant")
    print("Reject the Null Hypothesis.")
else:
    print("Result: Not Statistically Significant")
    print("Fail to Reject the Null Hypothesis.")
```

## Hypothesis Test 3: Total Charges and Customer Churn

### Hypotheses

- Null Hypothesis (H₀): The average Total Charges are the same for churned and non-churned customers.
- Alternative Hypothesis (H₁): The average Total Charges are different for churned and non-churned customers.

### Result

The p-value of approximately 0.925 is much greater than the significance level of 0.05.

Therefore, we fail to reject the null hypothesis.

There is no statistically significant difference in Total Charges between churned and non-churned customers in this dataset.

This suggests that Total Charges alone may not be an important factor associated with customer churn in this analysis.

## 95% Confidence Interval for Monthly Charges

```python


monthly_charges = df["MonthlyCharges"]

mean = monthly_charges.mean()
std_error = stats.sem(monthly_charges)

confidence_interval = stats.t.interval(
    0.95,
    len(monthly_charges) - 1,
    loc=mean,
    scale=std_error
)

margin_of_error = (
    confidence_interval[1] - confidence_interval[0]
) / 2

print("========== 95% CONFIDENCE INTERVAL ==========")
print("Mean Monthly Charges:", mean)
print("Margin of Error:", margin_of_error)
print("Lower Bound:", confidence_interval[0])
print("Upper Bound:", confidence_interval[1])
print(
    f"95% Confidence Interval: "
    f"({confidence_interval[0]:.2f}, {confidence_interval[1]:.2f})"
)
```

## 95% Confidence Interval for Monthly Charges

The average Monthly Charges in the dataset are approximately 113.64.

The calculated margin of error is approximately ±4.55.

The 95% confidence interval for the average Monthly Charges is:

**(109.08, 118.19)**

Based on this sample, the underlying population mean Monthly Charges is estimated to lie within this interval with 95% confidence.
